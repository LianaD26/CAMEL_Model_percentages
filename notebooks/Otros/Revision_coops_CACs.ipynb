Revisar las cooperativas que no se encuentran de la lista, y hacer ajustes de algunos nombres:

```python
data["COPERATIVA  ESPECIALIZADA DE AHORRO Y CREDITO CANAPRO"] = data["COPERATIVA ESPECIALIZADA DE AHORRO Y CREDITO CANAPRO"].combine_first(data["COOPERATIVA ESPECIALIZADA DE AHORRO Y CREDITO CANAPRO"])
data["COOPERATIVA DE AHORRO Y CREDITO FINCOMERCIO LTDA"] = data["COOPERATIVA DE AHORRO Y CR??DITO FINCOMERCIO LTDA"].combine_first(data["COOPERATIVA DE AHORRO Y CRÉDITO FINCOMERCIO LTDA"])
data["COOPERATIVA JUDICIAL  DEL TOLIMA  LTDA"] = data["COOPERATIVA JUDICIAL DEL TOLIMA"].combine_first(data["COOPERATIVA JUDICIAL DEL TOLIMA LTDA"])
data["COOPERATIVA MULTIACTIVA DE EDUCADORES DE CASANARE  LTDA"] = data["COOPERATIVA MULTIACTIVA DE EDUCADORES DE CASANARE"].combine_first(data["COOPERATIVA MULTIACTIVA DE EDUCADORES DE CASANARE LTDA"])
data["ORGANIZACION COOPERATIVA DE AHORRO Y CREDITO NORTE DE BOYACA"] = data["ORGANIZACION COOPERATIVA DE AHORRO Y CR??DITO NORTE DE BOYACA"].combine_first(data["COOPERATIVA MULTIACTIVA DEL NORTE DE BOYACA"])
data["COOPERATIVA ANTIOQUEYA DE TRABAJADORES GRUPO CAFETERO"] = data["COOPERATIVA ANTIOQUEÑA DE TRABAJADORES GRUPO CAFETERO"].combine_first(data["COOPERATIVA ANTIOQUE¥A DE TRABAJADORES GRUPO CAFETERO"]).combine_first(data["COOPERATIVA ANTIOQUE??A DE TRABAJADORES GRUPO CAFETERO"])
data["PROGRESSA ENTIDAD COOPERATIVA DE AHORRO Y CREDITO"] = data["PROGRESSA ENTIDAD COOPERATIVA DE AHORRO Y CR??DITO"].combine_first(data["PROGRESSA ENTIDAD COOPERATIVA DE AHORRO Y CRÉDITO"])
```

```python
cacs.loc[
    cacs["NOMBREENTIDAD"] == "COPERATIVA  ESPECIALIZADA DE AHORRO Y CREDITO CANAPRO",
    "NOMBREENTIDAD"
] = "COOPERATIVA ESPECIALIZADA DE AHORRO Y CREDITO CANAPRO"

cacs.loc[
    cacs["NOMBREENTIDAD"] == "COOPERATIVA JUDICIAL  DEL TOLIMA  LTDA",
    "NOMBREENTIDAD"
] = "COOPERATIVA JUDICIAL DEL TOLIMA LTDA"

cacs.loc[
    cacs["NOMBREENTIDAD"] == "COOPERATIVA DE  TRABAJADORES DE LA INDUSTRIA MILITAR",
    "NOMBREENTIDAD"
] = "COOPERATIVA DE TRABAJADORES DE LA INDUSTRIA MILITAR"

cacs.loc[
    cacs["NOMBREENTIDAD"] == "COOPERATIVA DE AHORRO Y CREDITO VALLE DE SAN JOSE  LTDA",
    "NOMBREENTIDAD"
] = "COOPERATIVA DE AHORRO Y CREDITO VALLE DE SAN JOSE LTDA"
cacs.loc[
    cacs["NOMBREENTIDAD"] == "COOPERATIVA ANTIOQUEYA DE TRABAJADORES GRUPO CAFETERO",
    "NOMBREENTIDAD"
] = "COOPERATIVA ANTIOQUEÑA DE TRABAJADORES GRUPO CAFETERO"
```

```python
# Renombrar columnas específicas
data = data.rename(columns={'COOPERATIVA DE AHORRO Y CREDITO COOPSERVUNAL': 'COOPERATIVA DE AHORRO Y CREDITO COOSERVUNAL',
                            'COOPERATIVA DE AHORRO Y CR??DITO SANTA ROSA DE OSOS LIMITADA': 'COOPERATIVA DE AHORRO Y CREDITO SANTA ROSA DE OSOS LIMITADA',
                            'COOPERATIVA DE TRABAJADORES DE LAS EMPRESAS DE LA ORGANIZACI??N CARVAJAL': 'COOPERATIVA DE TRABAJADORES DE LAS EMPRESAS DE LA ORGANIZACION CARVAJAL',
                            'MULTIACTIVA EL ROBLE, ENTIDAD COOPERATIVA': 'EL ROBLE ENTIDAD COOPERATIVA',
                            'COOPERATIVA DE TRABAJADORES DE ENKA LTDA': 'COOPERATIVA ESPECIALIZADA DE AHORRO Y CREDITO COOPERENKA',
                            'COOPERATIVA DE AHORRO Y CREDITO DE LOS TRABAJADORES DEL SECTOR EDUCATIVO DE COLOMBIA LTDA': 'COOPERATIVA DE AHORRO Y CREDITO DE LOS TRABAJADORES DEL SECTOR EDUCATIVO'})
```

```python
# Nombres de cooperativas del primer dataframe
nombres_cooperativas = cacs["NOMBREENTIDAD"].dropna().tolist()

# Columnas del segundo dataframe
columnas_data = data.columns.tolist()

# Nombres que NO están en las columnas de data
no_encontradas = [
    nombre for nombre in nombres_cooperativas
    if nombre not in columnas_data
]

# Imprimirlas
for nombre in no_encontradas:
    print(nombre)
```

Eliminar esa cooperativa de la lista general porque no se encontró en el web scraping

```python
nombres_cooperativas.remove("COOPERATIVA MULTIACTIVA UNIVERSITARIA NACIONAL")
```

**Filtrar las cooperativas que tengan los años completos 2022-2025**

```python
cols_data_filt = nombres_cooperativas + ['CUENTA', 'NOMBRE CUENTA', 'DATE']
data_filt = data[cols_data_filt]
```

```python
data_filt = data_filt[data_filt["DATE"] >= "2022-01-01"]
data_filt
```

Llenar valor NaN con ceros

```python
data_filt.fillna(0, inplace=True)
```
